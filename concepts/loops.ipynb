```python
# for loop

for i in range(5):
    print(i)

print("")

for i in range(1, 6):
    print(i)



fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
    
```

```python
# While loop

i = 0

while i < 5:
    print(i)
    i += 1

    
```

```python
# Break and continue

for i in range(10):

    print(i)

    if i == 8:
        break


    if i == 7:
        continue
    
    print("This will not be printed when i is 7.")
```

```python
# Total of numbers from 1 to 10

total = 0

for i in range(1, 11):
    total += i

print("Total:", total)


# n * N+1 / 2

n = 10

total_formula = n * (n + 1) // 2

print("Total using formula:", total_formula)
```
