```python
a = 10
b = 5

# Arithmetic Operators
print("The sum of a and b is:", a + b)
print("The difference of a and b is:", a - b)
print("The product of a and b is:", a * b)
print("The quotient of a and b is:", a / b)
print("The remainder of a divided by b is:", a % b)
print("The floor division of a and b is:", a // b)
print("The exponentiation of a to the power of b is:", a ** b)
```

```python
x = 15


y = 4

# Comparison Operators
print (x == y)  # False
print (x != y)  # True
print (x > y)   # True
print (x < y)   # False
```

```python
age = 36

# Logical Operators
print("Is age greater than 18?", age > 18)  # True


print(age > 10 or age < 20)  # True

print(not(age > 10 ))  # False
```

```python
#Assignment Operators

score = 85
print(score)  # 85

score += 15
print(score)  # 100

score -= 15
print(score)  # 85

```
