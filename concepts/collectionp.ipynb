```python
## 1.List

weekdays = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]

print("Weekdays:")
print(type(weekdays))



weekdays.append("Saturday")

print("After appending Saturday:")
print(weekdays)

weekdays.insert(0, "Sunday")
print("After inserting Sunday at the beginning:")
print(weekdays)

weekdays[6]="Funday"
print("After changing the last element:")
print(weekdays)

```

```python
# Tuples are immutable, meaning their elements cannot be changed after creation. They are defined using parentheses ().


leap_years = (2020, 2024, 2028, 2032)
print("Leap Years:")
print(type(leap_years))

#leap_years[0] = 2021  # This will raise an error because tuples are immutable

leap_years = leap_years + (2036,)  # Add 2036 as a single-element tuple

print("After attempting to modify the tuple:")
print(leap_years)


```

```python
#3.Set 

prime_numbers = {2, 3, 5, 7, 11,11,13, 17, 19}

print("Prime Numbers:")
print(type(prime_numbers))
print("After adding duplicate 11:")
print(prime_numbers)  # Duplicates are automatically removed

prime_numbers.add(23)
print("After adding 23:")   
print(prime_numbers)
```

```python
#4.Dictionary

employee = {
    "name": "John Doe",
    "age": 30,
    "department": "IT"} 


print("Employee Dictionary:")
print(type(employee))
print(employee)
employee["age"] = 31  # Update age
print("After updating age:")
print(employee)

employee["position"] = "Developer"  # Add new key-value pair
print("After adding position:")
print(employee)

employee.pop("department")  # Remove department
print("After removing department:") 
print(employee)

employee.clear()  # Clear all items
print("After clearing the dictionary:") 

employee = {
    "name": "John Doe",
    "age": 30,

    "department": "IT"}     

print("After reinitializing the dictionary:")
print(employee)

employee.update({"age": 32, "position": "Manager"})  # Update multiple key-value pairs
print("After updating age and adding position:")
print(employee)


employee.setdefault("salary", 50000)  # Add salary if it doesn't exist
print("After setting default salary:") 
print(employee)

employee["salary"] = 60000  # Update salary

print(employee)
employee.popitem()  # Remove the last inserted item
```

```python
employees = [
    {"name": "John Doe", "age": 30, "department": "IT"},
    {"name": "Jane Smith", "age": 28, "department": "HR"}
]

# Add an employee
employees.append({
    "name": "David Lee",
    "age": 35,
    "department": "Finance"
})

# Display all employees
for employee in employees:
    print(employee["name"], employee["department"])

# Update an employee
employees[0]["age"] = 31

# Remove an employee
employees.pop(1)

print(employees)
```
