```python
# Conditional Statements in Python
age = 25

if age >= 18:
    print("You are an adult.")
else:
    print("You are not an adult.")


if age < 13:
    print("You are a child.")
elif age < 20:
    print("You are a teenager.") 
elif age < 65:
    print("You are an adult.")
else:
    print("You are a senior citizen.")



temperature = 31
if temperature > 30:
    print("It's a hot day.")

    
```

```python
# Nested if statements
number = 12

if number > 0:
    if number % 2 == 0:
        print("The number is a positive even integer.")
    else:
        print("The number is a positive odd integer.")
else:
    print("The number is not a positive integer.")

```

```python
# Largest of three numbers using nested if statements
num1 = 10
num2 = 20
num3 = 15

if num1 >= num2:
    if num1 >= num3:
        largest = num1
    else:
        largest = num3
else:
    if num2 >= num3:
        largest = num2
    else:
        largest = num3

print("The largest number is:", largest)
```
