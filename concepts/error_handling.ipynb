```python

# Error handling for the script

a = 9
b = 0

try:
    result = a / b
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")

```

```python
# What are the different types of errors in Python?
# These examples catch each error so the cell can run from beginning to end.

# 1. SyntaxError: The code does not follow Python's syntax rules.
try:
    compile("if True print('hello')", "<example>", "exec")
except SyntaxError as error:
    print("SyntaxError:", error)

# 2. TypeError: An operation is used with an incompatible type.
try:
    result = "5" + 5
except TypeError as error:
    print("TypeError:", error)

# 3. ValueError: The type is right, but the value is not acceptable.
try:
    number = int("hello")
except ValueError as error:
    print("ValueError:", error)

# 4. NameError: A name is used before it has been defined.
try:
    eval("undefined_variable")
except NameError as error:
    print("NameError:", error)

# 5. IndexError: A sequence index is outside its valid range.
try:
    colors = ["red", "green"]
    print(colors[5])
except IndexError as error:
    print("IndexError:", error)

# 6. KeyError: A dictionary key does not exist.
try:
    person = {"name": "Ava"}
    print(person["age"])
except KeyError as error:
    print("KeyError:", error)

# 7. AttributeError: An object does not have the requested attribute.
try:
    number = 42
    number.append(1)
except AttributeError as error:
    print("AttributeError:", error)

# 8. ZeroDivisionError: A number is divided by zero.
try:
    result = 9 / 0
except ZeroDivisionError as error:
    print("ZeroDivisionError:", error)
```

```python
# Example of try-except-finally in Python
try:
    numerator = 10
    denominator = 0
    result = numerator / denominator
except ZeroDivisionError as error:
    print("Exception caught:", error)
finally:
    print("The finally block always runs.")
```

```python
try:
    result = 10 / 0
except Exception as error:
    print("Exception type:", type(error).__name__)
    print("Message:", error)
```
