```python
# File Handling in Python : Create file , Read and Write to file


# Write to a file

# w - means write mode. It will create a new file if it does not exist
#  or overwrite the existing file if it exists.


with open("example.txt", "w") as file:
    file.write("Hello, World!")





```

```python
# file Overwrite example
with open("example.txt", "w") as file:
    file.write("This will overwrite the existing content.")
```

```python
# file read example
with open("example.txt", "r") as file:
    content = file.read()
    print(content)  
```

```python
# file append example
with open("example.txt", "a") as file:
    file.write("\nThis will append to the existing content.")   
```

```python
# file read line by line example
with open("example.txt", "r") as file:
    for line in file:
        print(line.strip())  # strip() removes the newline character at the end of each line            
        
```
